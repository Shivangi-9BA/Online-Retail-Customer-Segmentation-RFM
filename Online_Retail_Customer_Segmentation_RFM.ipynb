## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Make a copy of the original dataset
df_clean = df.copy()

# 1. Remove duplicate records
df_clean = df_clean.drop_duplicates()

# 2. Remove transactions with missing CustomerID
# CustomerID is required for customer segmentation
df_clean = df_clean.dropna(subset=['CustomerID'])

# 3. Remove cancelled transactions
# Invoice numbers starting with 'C' represent cancellations
df_clean = df_clean[
    ~df_clean['InvoiceNo'].astype(str).str.startswith('C')
]

# 4. Keep only positive quantities
df_clean = df_clean[df_clean['Quantity'] > 0]

# 5. Keep only positive unit prices
df_clean = df_clean[df_clean['UnitPrice'] > 0]

# 6. Calculate Revenue
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']

# 7. Convert CustomerID to integer
df_clean['CustomerID'] = df_clean['CustomerID'].astype(int)

# Check the cleaned dataset
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
````

***Data Wrangling – Manipulations and Insights***
The raw dataset was cleaned and prepared for further analysis and customer segmentation.

First, duplicate records were removed to prevent repeated transactions from affecting the analysis. Records with missing CustomerID were excluded because customer identification is essential for building customer-level segments.

Cancelled transactions were removed because they do not represent completed sales. Transactions with negative quantities were also excluded, as they generally represent returns or adjustments rather than actual purchases for the purpose of customer purchasing analysis. Transactions with zero or negative UnitPrice were removed because they do not represent valid positive-priced sales.

A new Revenue variable was created using:

Revenue = Quantity × UnitPrice

Finally, CustomerID was converted into an integer format for easier customer-level analysis.

These steps resulted in a cleaner dataset that is more suitable for exploratory analysis,RFM feature engineering, and unsupervised customer segmentation.

(Negative quantities were investigated as exceptional transactions and were excluded from the customer-purchase analysis because they generally represent returns or adjustments rather than completed sales.)

#### Chart 6 - RFM Feature Distribution

````python
# Create RFM Analysis Dataset

# Set the analysis date as one day after the last transaction
analysis_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (analysis_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',                                  # Frequency
    'Revenue': 'sum'                                         # Monetary
})

# Rename columns
rfm.columns = ['Recency', 'Frequency', 'Monetary']

# Display the RFM dataset
rfm.head()
````

````python
# Create a copy of RFM data
rfm_log = rfm.copy()

# Apply log transformation to reduce skewness
rfm_log['Recency'] = np.log1p(rfm_log['Recency'])
rfm_log['Frequency'] = np.log1p(rfm_log['Frequency'])
rfm_log['Monetary'] = np.log1p(rfm_log['Monetary'])

# Check the transformed data
rfm_log.head()
````

````python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Fit and transform RFM data
rfm_scaled = scaler.fit_transform(rfm_log)

# Convert back to DataFrame
rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=['Recency', 'Frequency', 'Monetary'],
    index=rfm.index
)

# Check scaled data
rfm_scaled.head()
````

````python
rfm_scaled.describe()
````

We're going to find the appropriate number of clusters using: Elbow Method

Looks at how clustering error decreases as the number of clusters increases.

````python
# Elbow Method

from sklearn.cluster import KMeans

inertia = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))

plt.plot(K_range, inertia, marker='o')

plt.title('Elbow Method for Optimal Number of Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Within-Cluster Sum of Squares (Inertia)')

plt.xticks(K_range)
plt.tight_layout()
plt.show()
````

Silhouette Score

Measures how well separated the resulting clusters are.

````python
# Silhouette Score for different numbers of clusters

from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

silhouette_scores = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(rfm_scaled)

    score = silhouette_score(rfm_scaled, labels)
    silhouette_scores.append(score)

# Display scores
for k, score in zip(K_range, silhouette_scores):
    print(f"K = {k}: Silhouette Score = {score:.3f}")
````

Now create the final K-Means model-

````python
# Final K-Means Model

kmeans_final = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

rfm['Cluster'] = kmeans_final.fit_predict(rfm_scaled)

rfm.head()
````

````python
# Cluster Profiling

cluster_profile = rfm.groupby('Cluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': 'mean',
    'Cluster': 'count'
})

cluster_profile.rename(
    columns={'Cluster': 'CustomerCount'},
    inplace=True
)

cluster_profile
````

The clustering analysis identified two distinct customer segments. Approximately 38% of customers belong to the high-value segment, characterized by more recent purchases, higher purchase frequency, and substantially higher monetary value. Their average monetary contribution per customer is approximately nine times that of the lower-engagement segment, highlighting the importance of customer retention

````python
# Scale RFM features for clustering and PCA

from sklearn.preprocessing import StandardScaler

rfm_features = ['Recency', 'Frequency', 'Monetary']

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm[rfm_features])

print("RFM data scaled successfully.")
print("Shape:", rfm_scaled.shape)
````

````python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

rfm_pca = pca.fit_transform(rfm_scaled)

cluster_visual = pd.DataFrame({
    'PCA1': rfm_pca[:, 0],
    'PCA2': rfm_pca[:, 1],
    'Cluster': kmeans_final.labels_
})

cluster_visual.head()
````

````python
# Final Customer Segmentation Visualization

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=cluster_visual,
    x='PCA1',
    y='PCA2',
    hue='Cluster',
    palette='deep',
    alpha=0.7,
    s=70
)

plt.title('Final Customer Segmentation Using K-Means')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Customer Segment')

plt.tight_layout()
plt.show()
````

### 2. Handling Outliers

````python
# Handling Outliers using IQR Capping

import numpy as np

# Numerical columns where extreme values can affect the analysis
outlier_columns = ['Quantity', 'UnitPrice']

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap extreme values instead of removing them
    df[col] = np.clip(df[col], lower_bound, upper_bound)

print("Outlier treatment using IQR capping completed.")

# Check the updated ranges
print("\nUpdated numerical ranges:")
print(df[['Quantity', 'UnitPrice']].describe())
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The Interquartile Range (IQR) method was used to identify extreme values in the numerical variables Quantity and UnitPrice. Instead of removing these observations, I used IQR capping to limit extreme values to the calculated lower and upper bounds. This approach helps reduce the influence of extreme transactions while preserving genuine customer purchase records, including potentially valuable high-spending customers.

### ML Model - 1

````python
# ML Model - 1 Implementation: K-Means Clustering

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Fit the K-Means model
kmeans_final = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

# Predict cluster labels
rfm['Cluster'] = kmeans_final.fit_predict(rfm_scaled)

# Calculate evaluation metric
silhouette = silhouette_score(rfm_scaled, rfm['Cluster'])

print("K-Means clustering completed successfully.")
print("Number of clusters:", 2)
print("Silhouette Score:", round(silhouette, 4))
````

Two clusters were selected based on the clustering analysis, representing lower-engagement and high-value customer segments. Model performance was evaluated using the Silhouette Score, where a higher score indicates better-defined and more separated clusters.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Hyperparameter Tuning for K-Means using Silhouette Score

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_values = range(2, 8)
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(rfm_scaled)
    score = silhouette_score(rfm_scaled, labels)
    silhouette_scores.append(score)

# Display results
for k, score in zip(k_values, silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")

# Select the best K
best_k = k_values[silhouette_scores.index(max(silhouette_scores))]

print("\nBest number of clusters:", best_k)
print("Best Silhouette Score:", round(max(silhouette_scores), 4))
````

##### Which hyperparameter optimization technique have you used and why?

I used a manual hyperparameter search over the number of clusters (K) and evaluated each value using the Silhouette Score. I tested K values from 2 to 7 and selected the value with the highest Silhouette Score. This approach was chosen because K-Means requires the number of clusters to be specified in advance, and the Silhouette Score helps identify the K that provides the best-defined and well-separated customer segments.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

Yes. Hyperparameter tuning showed that K = 2 provides the best clustering performance, with a Silhouette Score of 0.4328. The score decreased for higher values of K, reaching 0.3092 at K = 7. Therefore, K = 2 was selected as the optimal number of clusters because it produced the highest Silhouette Score and the most clearly separated customer segments.
